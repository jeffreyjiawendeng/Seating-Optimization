```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('seats.csv')

x_data = df['Cleanliness']
y_data = df['Noise']
plt.scatter(x_data, y_data)
plt.xlabel('Cleanliness')
plt.xlim(0, 100)
plt.ylabel('Noise')
plt.ylim(0, 100)
plt.title('Seat Attributes')
plt.grid(True)
plt.show()
```
